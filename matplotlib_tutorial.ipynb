```python
import matplotlib.pyplot as plt
```

```python
oscar_movies=[
    "The Dark Knight",
    "The Murt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
year=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232]
plt.plot(year,oscar_revenue)
plt.title("Oscar movies revenue in each year.")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.show()
```

```python
oscar_movies=[
    "The Dark Knight",
    "The Murt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
non_oscar_movies=["Avatar","Mona", "Final Destination","The Terminator","The Legend"]
year=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232]
non_oscar_revenue=[2000,3500,245,1234,567]
plt.plot(year,oscar_revenue,label="Oscar movies")
plt.plot(year,non_oscar_revenue, label="Non oscar movies")
plt.title("Oscar movies revenue in each year.")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()
plt.show()
```
