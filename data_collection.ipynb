# Web Scraping Activity

## From quotes, extract all quotes and authors from all pages with the life tag.

```python
import requests
from bs4 import BeautifulSoup

# Fetch data and store it

page_count=1

while True:
    URL=f"http://quotes.toscrape.com/page/{page_count}/"
    res=requests.get(URL)

    soup=BeautifulSoup(res.text,"lxml")
    quotes=soup.select("div.quote")

    if not quotes:
        print("No valid pages anymore..")
        break

    with open(f"scrape_data/quotes{page_count}.html","w",encoding="utf-8") as f:
        f.write(res.text)
        print(f"Downloaded data from page{page_count}.")
    page_count+=1
```

```python
# Extract useful information

life_quotes=[]
for page in range(1,11):
    with open(f"scrape_data/quotes{page}.html","r",encoding="utf-8") as f:
        html_content=f.read()
    soup=BeautifulSoup(html_content,"lxml")
    all_quotes=soup.select("div.quote")
    for q in all_quotes:
        all_tags=[]

        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())
        if "life" in all_tags:
            text=q.select_one("span.text").get_text()
            author=q.select_one("small.author").get_text()
            life_quotes.append([text,author])
life_quotes
```

```python
import pandas as pd
df=pd.DataFrame(life_quotes,columns=["Quote","Author"])
df
```

```python
df.to_csv("cleaned_data/data1.csv",index=False)
```
